#Loading data from GitHub

### 21. Find a CSV file in a GitHub repository.

22.Copy the RAW link of the CSV file.

23.Load the CSV into Colab using pd.read_csv(raw_url).

24.Print the dataset shape (rows, columns).

25.Plot one column from the dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 21 & 22: GitHub RAW CSV link (Gapminder dataset)
raw_url = "https://raw.githubusercontent.com/resbaz/r-novice-gapminder-files/master/data/gapminder-FiveYearData.csv"

# Step 23: Load CSV into Colab
df = pd.read_csv(raw_url)

# Step 24: Print dataset shape (rows, columns)
print("Dataset shape (rows, columns):", df.shape)

# Step 25: Plot continent distribution
continent_counts = df["continent"].value_counts()
continent_counts.plot(kind="bar", title="Distribution by Continent (Gapminder Dataset)")
plt.xlabel("Continent")
plt.ylabel("Number of Records")
plt.show()
```
